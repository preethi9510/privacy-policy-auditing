```python
import pandas as pd
from datasets import Dataset

# Load my labeled CSV from the Colab file storage
df = pd.read_csv("IoT_labeled_COPPA.csv")

# Define a prompt style that tells the model exactly what to look for
prompt_style = """Act as a Senior Privacy Compliance Auditor.
Analyze the following sentence from an IoT Privacy Policy.

### DECISION CRITERIA:
Label as 1 if: It mentions data collection from children, parental consent, or safety rights for minors.
Label as 0 if: It is general boilerplate, website navigation, or geography-specific law (GDPR/CCPA).

### Sentence:
{}

### Label:
{}"""

def formatting_prompts_func(examples):
    sentences = examples["text"]
    labels    = examples["coppa_label"]
    texts = []
    for sentence, label in zip(sentences, labels):
        # We fill in the template and add the EOS (End of String) token
        text = prompt_style.format(sentence, label) + tokenizer.eos_token
        texts.append(text)
    return { "text" : texts, }

# Convert the pandas dataframe to a Hugging Face Dataset
dataset = Dataset.from_pandas(df)
dataset = dataset.map(formatting_prompts_func, batched = True)

print(f"Dataset prepared with {len(dataset)} examples.")
```

```python
from trl import SFTTrainer
from transformers import TrainingArguments
from unsloth import is_bfloat16_supported

trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = dataset,
    dataset_text_field = "text",
    max_seq_length = 2048,
    args = TrainingArguments(
        per_device_train_batch_size = 4,
        gradient_accumulation_steps = 4,
        max_steps = 1200,        # Increased from 840 (gives more time for nuance)
        learning_rate = 5e-5,    # Slower learning for higher precision
        weight_decay = 0.05,     # Prevents "overfitting" on specific IoT device names
        lr_scheduler_type = "cosine", # Gradually slows down learning for a better finish
        fp16 = not is_bfloat16_supported(),
        bf16 = is_bfloat16_supported(),
        logging_steps = 1,
        optim = "adamw_8bit",
        seed = 3407,
    ),
)

# Starting the training process..
trainer_stats = trainer.train()
```

```python
import pandas as pd
from unsloth import FastLanguageModel

# Inference Mode
FastLanguageModel.for_inference(model)

prompt_style = """Act as a Senior Privacy Compliance Auditor.
Analyze the following sentence from an IoT Privacy Policy.

### DECISION CRITERIA:
Label as 1 if: It mentions data collection from children, parental consent, or safety rights for minors.
Label as 0 if: It is general boilerplate, website navigation, or geography-specific law (GDPR/CCPA).

### Sentence:
{}

### Label:
{}"""

def ai_label_sentence(sentence):
    if not isinstance(sentence, str) or len(sentence.strip()) < 5:
        return 0
    inputs = tokenizer([prompt_style.format(sentence, "")], return_tensors = "pt").to("cuda")
    outputs = model.generate(**inputs, max_new_tokens = 2, use_cache = True)
    decoded_output = tokenizer.batch_decode(outputs)[0]
    return 1 if "### Label:\n1" in decoded_output else 0

print("Loading dataset...")
file_path = "IoT_training_needLabeling.csv"

try:
    # Checking for the binary 'PK' signature
    with open(file_path, 'rb') as f:
        is_binary = f.read(2) == b'PK'

    if is_binary:
        print("Binary data detected! Converting Excel to clean text...")
        df = pd.read_excel(file_path)
    else:
        df = pd.read_csv(file_path, encoding="latin1", engine='python', on_bad_lines='skip')

    # Find the text column automatically
    target_col = df.select_dtypes(include=['object']).apply(lambda x: x.str.len().mean()).idxmax()
    df = df.rename(columns={target_col: 'text'})

    # Remove any remaining rows that are pure gibberish/empty
    df = df[df['text'].str.contains('[a-zA-Z]', na=False)]

except Exception as e:
    print(f"Load failed: {e}. Trying raw string recovery...")
    df = pd.read_csv(file_path, encoding="latin1", on_bad_lines='skip', header=None)
    df.columns = ['text'] + list(df.columns[1:])

# AI Model Labeling
print(f"Ready to label {len(df)} cleaned rows...")
df['coppa_label'] = df['text'].apply(ai_label_sentence)

# Saving (UTF-8-SIG for Excel compatibility)
df.to_csv("IoT_AI_Labeled_Results.csv", index=False, encoding='utf-8-sig')
print("Complete! Open 'IoT_AI_Labeled_Results.csv' in Excel.")
```

```python
import pandas as pd
import re

# Defining the original rule-based logic to compare against the AI model
def rule_based_label(text):
    if not isinstance(text, str): return 0
    text_str = text.lower().strip()
    BANNED_START = ('menu ', 'updated ', 'effective ', 'page ', 'section ', 'table of ', 'visit ', 'follow ', 'copyright ', 'portions of ')
    if text_str.startswith(BANNED_START) or len(text_str) < 45: return 0
    GEOGRAPHY_EXCLUSION = r"(eu |eea|switzerland|california|ccpa|gdpr|united kingdom|member state)"
    if re.search(GEOGRAPHY_EXCLUSION, text_str): return 0
    COPPA_SUBJECTS = r"(child|minor|under 13|under 18|age|birthday|parent|guardian|consent|verifi)"
    DATA_ACTIONS = r"(collect|use|share|disclose|right|request|contact|identif|provide|store|protect|remove|delete)"
    if text_str.endswith('?'): return 0
    if re.search(COPPA_SUBJECTS, text_str) and re.search(DATA_ACTIONS, text_str): return 1
    if re.search(DATA_ACTIONS, text_str) and len(text_str) > 80: return 1
    return 0

# Loading the AI model results
df = pd.read_csv("IoT_AI_Labeled_Results.csv", encoding="utf-8-sig")

# Calculate Rule-Based labels for comparison
df['rule_label'] = df['text'].apply(rule_based_label)

# Statistics
matches = (df['coppa_label'] == df['rule_label']).sum()
accuracy = (matches / len(df)) * 100
nuance_caught = ((df['rule_label'] == 0) & (df['coppa_label'] == 1)).sum()

print("="*40)
print(f"      ACCURACY VALIDATION REPORT")
print("="*40)
print(f"Total Sentences Analyzed: {len(df)}")
print(f"AI/Rule Agreement Rate:   {accuracy:.2f}%")
print(f"AI 'Nuance' Detections:   {nuance_caught}")
print("-"*40)
print("Note: 'Nuance' detections are sentences the AI labeled as COPPA-relevant")
print("that the basic Python script missed due to lack of specific keywords.")
print("="*40)
```

```python
# Create a 'Disagreement' dataframe
disagreements = df[df['coppa_label'] != df['rule_label']]

# See the top 10 rows where they disagreed
print("--- TOP DISAGREEMENTS ---")
for i, row in disagreements.head(10).iterrows():
    print(f"Text: {row['text'][:100]}...")
    print(f"AI said: {row['coppa_label']} | Script said: {row['rule_label']}\n")
```
